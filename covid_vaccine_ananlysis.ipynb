```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
```

```python
df = pd.read_csv("covid_vaccine_statewise.csv")

print("===== FIRST 5 ROWS =====")
display(df.head())
```

```python
print("===== DATASET INFO =====")
df.info()

print("\n===== STATISTICAL SUMMARY =====")
display(df.describe())

print("\n===== COLUMN NAMES =====")
print(df.columns)

print("\n===== MISSING VALUES =====")
display(df.isnull().sum())
```

```python
# Fill missing values
df.fillna(0, inplace=True)
```

```python
df.columns = df.columns.str.strip()

# rename columns to simple names
df.rename(columns={
    "Male (Doses Administered)": "Male",
    "Female (Doses Administered)": "Female",
    "First Dose": "First Dose",
    "Second Dose": "Second Dose"
}, inplace=True)

print(df.columns)  # check
```

```python
first_dose = df.groupby("State")["First Dose"].sum()

print("===== FIRST DOSE (STATE-WISE) =====")
display(first_dose)
```

```python
second_dose = df.groupby("State")["Second Dose"].sum()

print("\n===== SECOND DOSE (STATE-WISE) =====")
display(second_dose)
```

```python
print(df.columns)
```

```python
total_male = df["Male"].sum()

print("\n===== TOTAL MALES VACCINATED =====")
print(total_male)
```

```python
total_female = df["Female"].sum()

print("\n===== TOTAL FEMALES VACCINATED =====")
print(total_female)
```

```python
first_dose.to_csv("first_dose_statewise.csv")
second_dose.to_csv("second_dose_statewise.csv")

print("✅ Files saved successfully!")
```

```python
first_dose.sort_values(ascending=False).head(10).plot(kind='bar')
plt.title("Top 10 States - First Dose Vaccination")
plt.xlabel("State")
plt.ylabel("People Vaccinated")
plt.show()
```

```python
plt.pie([total_male, total_female],
        labels=["Male", "Female"],
        autopct="%1.1f%%")
plt.title("Male vs Female Vaccination")
plt.show()
```
